## Question 1 – What is Streamlit and why is it popular for data apps?
**Answer:** Streamlit is an open-source Python framework for building interactive web applications, especially data science, machine-learning, analytics, and visualization apps. It is popular because developers can create apps using Python without needing to write separate HTML, CSS, or JavaScript for basic interfaces.

**Install:** `pip install streamlit`

**Script rerun model:** Streamlit reruns the script from top to bottom when a user interacts with a widget, such as clicking a button or changing a slider. It also reruns after the source file is edited and saved. Ordinary local variables are recreated on reruns; use `st.session_state` for values that must persist.

## Question 2 – Demonstrate Streamlit text and formatting elements
**Answer:** This program demonstrates a title, header, subheader, `st.write`, Markdown bullets, caption, code block, and divider.

````python
import streamlit as st

st.set_page_config(page_title="Text Elements Demo", layout="wide")
st.title("Streamlit Text Elements")
st.header("Introduction")
st.subheader("What is Streamlit?")
st.write("Streamlit helps us build interactive data apps using Python.")

st.markdown("""
### Main features
- Simple Python-based development
- Interactive widgets
- DataFrame and chart support
""")

st.caption("A small caption below the main content.")
st.code('''import streamlit as st
st.title("Hello Streamlit")
st.write("My first app")''', language="python")
st.divider()
st.write("The divider separates sections of the page.")
````

**Difference between `st.write` and `st.markdown`:** `st.write()` is a flexible display function that can display text, numbers, Python objects, DataFrames, and charts. `st.markdown()` renders text with Markdown formatting such as headings, bold text, links, and bullet lists.

## Question 3 – Student report card app
**Answer:** Use the supplied dictionary. `st.metric` displays the current marks and the delta (current marks minus previous marks). For marks = **82**, the delta is **+7** and `st.info` is displayed because 82 is in the 75–89 range.

````python
import streamlit as st

student = {"name": "Asha Verma", "course": "Python", "marks": 82,
           "previous_marks": 75}

st.title("Student Report Card")
st.write("Student Name:", student["name"])
st.write("Course:", student["course"])
delta = student["marks"] - student["previous_marks"]
st.metric("Current Marks", student["marks"], delta=delta)

marks = student["marks"]
if marks >= 90:
    st.success("Excellent performance!")
elif marks >= 75:
    st.info("Good performance!")
elif marks >= 50:
    st.warning("Needs improvement.")
else:
    st.error("Poor performance. More practice is needed.")

st.subheader("Student Data")
st.json(student)
````

## Question 4 – BMI calculator
**Answer:** The app uses `st.text_input` for the name, two `st.number_input` widgets for weight and height, `st.selectbox` for measurement preference, and `st.checkbox` for confirmation. BMI is `weight_kg / (height_m ** 2)`. Categories: Underweight below 18.5; Normal 18.5 to below 25; Overweight 25 to below 30; Obese 30 or above.

**Why does nothing appear before pressing Calculate?** The result code is inside the button condition. Before the click, the condition is false, so the calculation and output are not executed. Streamlit reruns the script after the click.

````python
import streamlit as st

st.title("BMI Calculator")
name = st.text_input("Enter your name")
weight = st.number_input("Weight (kg)", min_value=1.0, max_value=300.0, value=60.0)
height = st.number_input("Height (metres)", min_value=0.5, max_value=2.5, value=1.65)
unit = st.selectbox("Measurement system", ["Metric (kg and metres)"])
confirmed = st.checkbox("I confirm that the entered details are correct")

if st.button("Calculate"):
    if not name.strip():
        st.error("Please enter your name.")
    elif not confirmed:
        st.warning("Please confirm your details.")
    else:
        bmi = weight / (height ** 2)
        st.write(f"Hello, {name}!")
        st.metric("Your BMI", f"{bmi:.2f}")
        if bmi < 18.5:
            category = "Underweight"
        elif bmi < 25:
            category = "Normal"
        elif bmi < 30:
            category = "Overweight"
        else:
            category = "Obese"
        st.write("Category:", category)
````

## Question 5 – Layout tools and city sales dashboard
**Answer – layout tools:**
1. `st.sidebar`: places controls/content in a sidebar.
2. `st.columns`: arranges elements side by side.
3. `st.tabs`: organizes content into selectable tabs.
4. `st.expander`: creates collapsible content.
5. `st.container`: groups related elements.

The program filters a city DataFrame with a sidebar slider, displays three metrics in columns, and provides Table and Chart tabs plus an expander.

````python
import streamlit as st
import pandas as pd

st.title("City Sales Dashboard")
sales_data = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Bengaluru", "Mumbai", "Delhi"],
    "Sales": [85000, 62000, 95000, 78000, 55000],
    "Orders": [120, 95, 140, 110, 80],
    "Region": ["South", "South", "South", "West", "North"]
})

with st.sidebar:
    st.header("Filter")
    min_sales = st.slider("Minimum sales", 0, 95000, 50000, 5000)

filtered = sales_data[sales_data["Sales"] >= min_sales]
with st.container():
    st.subheader("Sales Summary")
    c1, c2, c3 = st.columns(3)
    c1.metric("Cities shown", len(filtered))
    c2.metric("Total sales", f"{filtered['Sales'].sum():,}")
    c3.metric("Total orders", int(filtered["Orders"].sum()))

table_tab, chart_tab = st.tabs(["Table", "Chart"])
with table_tab:
    st.dataframe(filtered, use_container_width=True)
with chart_tab:
    if not filtered.empty:
        st.bar_chart(filtered.set_index("City")[["Sales"]])
    else:
        st.info("No cities match the selected minimum sales.")

with st.expander("Dashboard description"):
    st.write("This dashboard shows sample sales and order data. Use the sidebar slider to filter cities by sales.")
````

## Question 6 – Course registration form with validation
**Answer:** `st.form` groups related inputs and submits them together. Validate that the name has at least 3 characters, the email contains `@` and a dot, and terms are accepted. Display every error using `st.error`; if there are no errors, show success and call `st.balloons()`.

**Why is a form better?** A form batches the values and sends them together on submission. Separate widgets normally cause reruns as each widget changes, which can trigger processing before all details are entered.

````python
import streamlit as st

st.title("Course Registration")
with st.form("registration_form"):
    name = st.text_input("Full name")
    email = st.text_input("Email address")
    course = st.selectbox("Choose a course", ["Python", "Data Science", "Web Development", "Machine Learning"])
    study_mode = st.radio("Study mode", ["Online", "Offline"])
    accepted_terms = st.checkbox("I accept the terms and conditions")
    submitted = st.form_submit_button("Register")

if submitted:
    errors = []
    if len(name.strip()) < 3:
        errors.append("Name must contain at least 3 characters.")
    if "@" not in email.strip() or "." not in email.strip():
        errors.append("Email must contain @ and a dot.")
    if not accepted_terms:
        errors.append("You must accept the terms and conditions.")
    if errors:
        for error in errors:
            st.error(error)
    else:
        st.success(f"Registration successful for {name.strip()} – {course} ({study_mode}).")
        st.balloons()
````

## Question 7 – `st.session_state` and counter app
**Answer:** `st.session_state` stores values across reruns for a user's session. It is needed to remember values such as counters and selections.

A plain local Python variable is recreated whenever Streamlit reruns the script, so it does not accumulate its old value. After three Increase clicks, the session-state counter displays **3**. The plain variable displays **1** on the third click because it is reset to zero on each rerun and incremented once for that click.

````python
import streamlit as st

st.title("Counter App")
if "count" not in st.session_state:
    st.session_state.count = 0

plain_count = 0
col1, col2, col3 = st.columns(3)

if col1.button("Increase"):
    st.session_state.count += 1
    plain_count += 1
if col2.button("Decrease"):
    st.session_state.count = max(0, st.session_state.count - 1)
    plain_count -= 1
if col3.button("Reset"):
    st.session_state.count = 0
    plain_count = 0

st.metric("Session state counter", st.session_state.count)
st.write("Plain Python variable:", plain_count)
````

## Question 8 – To-do list app using `st.session_state`
**Answer:** Tasks are stored in session state. A keyed text input uses an `on_change` callback to add the task and clear the input. Each task has a checkbox and Delete button with unique keys. Deletion uses a callback with `args`.

**Why unique keys in a loop?** Streamlit uses keys to distinguish widgets and preserve their values. Reusing a key for multiple widgets creates a duplicate-key error or ambiguity. Use a unique task ID in each widget key.

````python
import streamlit as st

st.title("To-Do List")
if "tasks" not in st.session_state:
    st.session_state.tasks = []
if "next_task_id" not in st.session_state:
    st.session_state.next_task_id = 1

def add_task():
    value = st.session_state.new_task.strip()
    if value:
        st.session_state.tasks.append({
            "id": st.session_state.next_task_id, "text": value, "done": False
        })
        st.session_state.next_task_id += 1
    st.session_state.new_task = ""

def delete_task(task_id):
    st.session_state.tasks = [t for t in st.session_state.tasks if t["id"] != task_id]

st.text_input("Enter a new task and press Enter", key="new_task", on_change=add_task)
st.button("Add Task", on_click=add_task)

for task in st.session_state.tasks:
    task_id = task["id"]
    c1, c2 = st.columns([4, 1])
    with c1:
        task["done"] = st.checkbox(task["text"], value=task["done"],
                                   key=f"task_checkbox_{task_id}")
    with c2:
        st.button("Delete", key=f"delete_button_{task_id}",
                  on_click=delete_task, args=(task_id,))

completed = sum(1 for t in st.session_state.tasks if t["done"])
total = len(st.session_state.tasks)
st.write(f"{completed} of {total} tasks completed")
````

## Question 9 – CSV explorer data app
**Answer:** The app uses `st.file_uploader`; if no file is uploaded it uses a built-in sample DataFrame. A function decorated with `@st.cache_data` reads the CSV. It displays `head()`, shape, and `describe()`, filters a text column using `selectbox` and `multiselect`, plots a numeric column with a bar chart, and provides a CSV download button.

**`st.cache_data` vs `st.cache_resource`:** `st.cache_data` is intended for data results such as DataFrames and transformed data. `st.cache_resource` is intended for shared resources such as database connections or machine-learning models that should be initialized once and reused.

````python
import streamlit as st
import pandas as pd

st.title("CSV Explorer")

@st.cache_data
def read_csv(uploaded_file):
    return pd.read_csv(uploaded_file)

uploaded_file = st.file_uploader("Upload a CSV file", type=["csv"])
if uploaded_file is not None:
    df = read_csv(uploaded_file)
else:
    df = pd.DataFrame({
        "City": ["Hyderabad", "Chennai", "Bengaluru", "Mumbai", "Delhi"],
        "Category": ["Electronics", "Books", "Electronics", "Furniture", "Books"],
        "Sales": [85000, 62000, 95000, 78000, 55000],
        "Orders": [120, 95, 140, 110, 80]
    })
    st.info("Using the built-in sample data.")

st.subheader("First rows")
st.dataframe(df.head(), use_container_width=True)
st.write("Shape (rows, columns):", df.shape)
st.subheader("Descriptive statistics")
st.write(df.describe(include="all"))

text_cols = df.select_dtypes(include=["object", "string", "category"]).columns.tolist()
numeric_cols = df.select_dtypes(include="number").columns.tolist()

if text_cols:
    text_col = st.selectbox("Choose a text column", text_cols)
    options = df[text_col].dropna().astype(str).unique().tolist()
    chosen = st.multiselect("Select values", options, default=options)
    filtered = df[df[text_col].astype(str).isin(chosen)]
else:
    filtered = df.copy()
    st.warning("No text columns are available.")

st.subheader("Filtered data")
st.dataframe(filtered, use_container_width=True)

if numeric_cols and not filtered.empty:
    numeric_col = st.selectbox("Choose a numeric column", numeric_cols)
    st.bar_chart(filtered.set_index(filtered.columns[0])[[numeric_col]])
elif not numeric_cols:
    st.warning("No numeric columns are available.")

csv_data = filtered.to_csv(index=False).encode("utf-8")
st.download_button("Download filtered data as CSV", csv_data,
                   file_name="filtered_data.csv", mime="text/csv")
````

## Question 10 – Deploy a Streamlit app on Streamlit Community Cloud

**1. Project folder structure**
```text
streamlit-project/
├── app.py
├── requirements.txt
├── .gitignore
└── .streamlit/
    └── secrets.toml
```

**2. `requirements.txt`**
```text
streamlit
pandas
```

**3. `.gitignore`**
```gitignore
.streamlit/secrets.toml
__pycache__/
*.pyc
.venv/
venv/
```

**4. Read a secret in Python**
```python
import streamlit as st
api_key = st.secrets["API_KEY"]
```

**5. Local `.streamlit/secrets.toml`**
```toml
API_KEY = "your-secret-api-key"
```
Replace the example with the actual secret locally.

**6. Git commands**
```bash
git init
git add app.py requirements.txt .gitignore
git commit -m "Add Streamlit app"
git branch -M main
git remote add origin https://github.com/USERNAME/REPOSITORY.git
git push -u origin main
```
Replace `USERNAME` and `REPOSITORY` with your GitHub details.

**7. Deployment steps**
1. Create a GitHub repository and push the project files.
2. Sign in to Streamlit Community Cloud using GitHub.
3. Select **Create app**.
4. Choose the repository, branch (usually `main`), and main file path (for example, `app.py`).
5. If the app needs secrets, add them in the app's **Secrets** settings using TOML format.
6. Deploy, wait for the build to finish, and open the generated app URL.

**Why must `secrets.toml` never be committed?** It may contain API keys, passwords, database credentials, or tokens. Committing it can expose credentials, allowing others to misuse services or access data. Add it to `.gitignore` and configure production secrets in Streamlit Community Cloud's Secrets settings.
